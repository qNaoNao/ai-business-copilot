```python
def compare_category(category_name):
    category_data = (
        jul_aug_sales[
            jul_aug_sales["category"] == category_name
        ]
        .groupby("month")
        .agg(
            gmv=("price", "sum"),
            orders=("order_id", "nunique"),
            customers=("customer_unique_id", "nunique"),
            items=("order_item_id", "count"),
            avg_item_price=("price", "mean")
        )
    )

    category_data["aov"] = (
        category_data["gmv"] / category_data["orders"]
    )

    category_data["avg_items_per_order"] = (
        category_data["items"] / category_data["orders"]
    )

    july = category_data.loc[
        pd.Period("2018-07", freq="M")
    ]

    august = category_data.loc[
        pd.Period("2018-08", freq="M")
    ]

    comparison = pd.DataFrame({
        "July": july,
        "August": august,
        "Change %": ((august / july) - 1) * 100
    })

    return comparison.round(2)
```

```python
def get_monthly_kpis(start_month, end_month, conn):
    query = """
    SELECT
        SUBSTR(o.order_purchase_timestamp, 1, 7) AS month,
        ROUND(SUM(f.product_value), 2) AS gmv,
        COUNT(DISTINCT o.order_id) AS orders,
        COUNT(DISTINCT c.customer_unique_id) AS customers,
        SUM(f.item_count) AS items,
        ROUND(SUM(f.freight_value), 2) AS freight
    FROM orders AS o

    JOIN customers AS c
        ON o.customer_id = c.customer_id

    JOIN order_financials AS f
        ON o.order_id = f.order_id

    WHERE o.order_status = 'delivered'
      AND SUBSTR(o.order_purchase_timestamp, 1, 7)
          BETWEEN ? AND ?

    GROUP BY month
    ORDER BY month;
    """

    result = pd.read_sql_query(
        query,
        conn,
        params=[start_month, end_month]
    )

    result["aov"] = (
        result["gmv"] / result["orders"]
    )

    result["avg_items_per_order"] = (
        result["items"] / result["orders"]
    )

    result["avg_item_price"] = (
        result["gmv"] / result["items"]
    )

    result["freight_ratio"] = (
        result["freight"]
        / (result["gmv"] + result["freight"])
    )

    return result
```

```python
def get_category_contribution(
    base_month,
    compare_month,
    conn
):
    query = """
    SELECT
        COALESCE(
            p.product_category_name_english,
            p.product_category_name,
            'unknown'
        ) AS category,

        ROUND(SUM(
            CASE
                WHEN SUBSTR(
                    o.order_purchase_timestamp, 1, 7
                ) = ?
                THEN oi.price
                ELSE 0
            END
        ), 2) AS base_gmv,

        ROUND(SUM(
            CASE
                WHEN SUBSTR(
                    o.order_purchase_timestamp, 1, 7
                ) = ?
                THEN oi.price
                ELSE 0
            END
        ), 2) AS compare_gmv

    FROM orders AS o

    JOIN order_items AS oi
        ON o.order_id = oi.order_id

    JOIN products AS p
        ON oi.product_id = p.product_id

    WHERE o.order_status = 'delivered'
      AND SUBSTR(
          o.order_purchase_timestamp, 1, 7
      ) IN (?, ?)

    GROUP BY category;
    """

    result = pd.read_sql_query(
        query,
        conn,
        params=[
            base_month,
            compare_month,
            base_month,
            compare_month
        ]
    )

    result["gmv_change"] = (
        result["compare_gmv"]
        - result["base_gmv"]
    )

    result["change_pct"] = (
        result["gmv_change"]
        / result["base_gmv"]
        * 100
    )

    return result.sort_values("gmv_change")
```

```python
def compare_category(
    category_name,
    base_month,
    compare_month,
    conn
):
    query = """
    SELECT
        SUBSTR(o.order_purchase_timestamp, 1, 7) AS month,

        ROUND(SUM(oi.price), 2) AS gmv,

        COUNT(DISTINCT o.order_id) AS orders,

        COUNT(DISTINCT c.customer_unique_id) AS customers,

        COUNT(*) AS items,

        ROUND(AVG(oi.price), 2) AS avg_item_price

    FROM orders AS o

    JOIN customers AS c
        ON o.customer_id = c.customer_id

    JOIN order_items AS oi
        ON o.order_id = oi.order_id

    JOIN products AS p
        ON oi.product_id = p.product_id

    WHERE o.order_status = 'delivered'

      AND COALESCE(
            p.product_category_name_english,
            p.product_category_name,
            'unknown'
          ) = ?

      AND SUBSTR(
            o.order_purchase_timestamp, 1, 7
          ) IN (?, ?)

    GROUP BY month
    ORDER BY month;
    """

    result = pd.read_sql_query(
        query,
        conn,
        params=[
            category_name,
            base_month,
            compare_month
        ]
    )

    result["aov"] = (
        result["gmv"] / result["orders"]
    )

    result["avg_items_per_order"] = (
        result["items"] / result["orders"]
    )

    result = result.set_index("month")

    if (
        base_month not in result.index
        or compare_month not in result.index
    ):
        raise ValueError(
            "The category has no data in one of the selected months."
        )

    base = result.loc[base_month]
    compare = result.loc[compare_month]

    comparison = pd.DataFrame({
        "Base": base,
        "Compare": compare,
        "Change %": (
            (compare / base) - 1
        ) * 100
    })

    return comparison
```
